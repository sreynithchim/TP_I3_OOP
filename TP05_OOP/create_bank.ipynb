```python
"""Create a class named BankAccount"""
class BankAccount:
    
    bank_name = "ITC Bank"
    """Constructor with owner, account_number, and balance"""
    def __init__(self, owner, account_number, balance):
        self.owner = owner
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited: ${amount}")
        else:
            print("Deposit amount must be positive.")

    
    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive.")
        elif amount > self.balance:
            print("Insufficient balance.")
        else:
            self.balance -= amount
            print(f"Withdrew: ${amount}")


    def display_balance(self):
        print(f"Current Balance: ${self.balance}")


"""Create two bank account objects"""
account1 = BankAccount("Visal", "ACC001", 1000)
account2 = BankAccount("Sokha", "ACC002", 500)

"""Test deposit and withdrawal operations"""
print(f"Bank: {BankAccount.bank_name}")
print(f"Account Owner: {account1.owner}")
print(f"Initial Balance: ${account1.balance}")

account1.deposit(500)
account1.display_balance()

account1.withdraw(300)
account1.display_balance()

"""Testing the requirements"""""
account1.deposit(-100)     
account1.withdraw(-50)      
account1.withdraw(5000)     
account1.display_balance()

print()
print(f"Account Owner: {account2.owner}")
print(f"Initial Balance: ${account2.balance}")
account2.deposit(200)
account2.withdraw(100)
account2.display_balance()
```

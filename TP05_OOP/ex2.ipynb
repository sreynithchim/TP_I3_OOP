```python
class Dog:
    species = "Canine"
    def __init__(self,name,kind):
        self.name=name
        self.kind=kind 
    
        
dog1 = Dog("Jerry", "Golden Retriever") 
dog2 = Dog("JoJo", "Pomeranian")
dog3 = Dog("Max", "German Shepherd")


for dog in [dog1, dog2, dog3]:
    print("Species :", dog.species)
    print("Name :", dog.name)
    print("Kind :", dog.kind)
    print()

dog1.name = "Rex"

print("Updated information")
print("Species :", dog1.species)
print("Name :", dog1.name)
print("Kind :", dog1.kind)
```

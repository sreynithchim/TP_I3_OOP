```python
class BankCustomer:
    def __init__(self,name ,customer_id,balance):
        
        self.__name=name
        self.__customer_id=customer_id
        self.__balance=balance if balance>=0 else 0
    
    def get_name(self):
        return self.__name
    
    def get_customer_id(self):
        return self.__customer_id
    
    def get_balance(self):
        return self.__balance
    
    def deposit(self,amount):
        if amount<=0:
            print("Deposite must be positive.")
        else:
            self.__balance+=amount
    def withdrawal(self,amount):
        if amount<=0:
            print("Withdrawal must be positive")
        else:
            self.__balance-=amount
    def transfer(self,other,amount):
        if other.get_customer_id() == self.__customer_id:
            print("Cannot transfer to the same account.")
            return False
   
        if self.withdraw(amount):
            other.deposit(amount)
            return True
        return False
            
    def __str__(self):
        return (f"Name : {self.__name}\n"
                f"Customer_ID : {self.__customer_id}\n"
                f"Balance : {self.__balance}$")

visal = BankCustomer("Visal", "C001", 2000)
dara = BankCustomer("Dara", "C002", 1500)
sna = BankCustomer("Sna", "C003", 1000)




print("========== BANK CUSTOMER SYSTEM ==========")
print()
print(f"Customer: {visal.get_name()}")
print(f"Customer ID: {visal.get_customer_id()}")
print(f"Initial Balance: {visal.get_balance()}")
print()

print("Deposit: 500")
visal.deposit(500)
print(f"New Balance: {visal.get_balance()}")

print("========== TRANSACTION COMPLETED ==========")

print("--- Extra tests ---")
sna.withdrawal(400)          # works, Sna now has 600
sna.withdrawal(5000)         # Insufficient balance
visal.deposit(-50)         # Invalid deposit amount
visal.transfer(visal, 100) # Cannot transfer to the same account
print(f"Dara: {dara.get_balance()}")  # 1800
print(f"Sna: {sna.get_balance()}")    # 600
        
    
        
```

```python
class Course:
    def __init__(self,course_name,course_code):
        self.__course_name = course_name
        self.__course_code=course_code
        self.__students=[]
        
    def add_student(self,name):
        if name in self.__students:
            print(f"{name} already register")
        else:
            self.__students.append(name)
    def remove_student(self,name):
        if name in self.__students:
            self.__students.remove(name)
        else:
            print("This name does not exist.")
    def display_student(self):
        print("Course Name : ",self.__course_name)
        print("Course Code : ",self.__course_code)
        print("===================================")
        print("Registered Students")
        for number, name in enumerate(self.__students, start=1):
            print(f"{number}. {name}")
        
        
    def student_count(self):
        return len(self.__students)

course = Course("Advanced Programming OOP", "AMS_A")


course.add_student("Visal")
course.add_student("Dara")
course.add_student("Sna")
course.add_student("Lina")
course.add_student("Vanna")

course.display_student()
print()
print(f"Total Students: {course.student_count()}")
        
    
        
```

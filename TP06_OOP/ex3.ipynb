```python
class Rectangle:
    def __init__(self,length,width):
        self.__lenghth=length
        self.__width=width
        
    def get_length(self):
        return self.__lenghth
    
    def get_width(self):
        return self.__width
    
    def is_valid(self,length,width):
        if length<0 and width<0:
            print("Length and width must be positive.")
        else:
            return self.__lenghth,self.__width
    def area(self):
        return self.__width*self.__lenghth
    
    def perimeter(self):
        return (self.__width+self.__lenghth)*2
    def display(self):
        print("Length : ",self.__lenghth)
        print("Width : ",self.__width)
        print("Area : ",self.area())
        print("Perimeter : ",self.perimeter())
    def __str__(self):
        return  ("Length : ",self.__lenghth,
        "Width : ",self.__width,
        "Area : ",self.area(),
        "Perimeter : ",self.perimeter())
        
rect1=Rectangle(10,4)

rect1.display()
```

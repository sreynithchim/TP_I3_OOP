```python
class Mobile_Phone:
    def __init__(self,brand,model,battery):
        self.__brand=brand
        self.__model=model
        self.__battery=max(0,min(100,battery))
        
    def get_brand(self):
        return self.__brand
    def get_model(self):
        return self.__model
    def get_battery(self):
        return self.__battery
    
    def charge(self,amount):
        if amount>0:
            self.__battery=min(100,self.__battery+amount)
        
    def use_battery(self,amount):
        if amount>0:
            self.__battery=max(0, self.__battery - amount)
            
    def __str__(self):
        return (f"Brand : {self.__brand}\n"
                f"Model : {self.__model}\n"
                f"Battery : {self.__battery}%")
        
phone=Mobile_Phone("iPhone","16 Pro max",80)
print(phone)
print("Charging: 10%")
phone.charge(10)
print(f"New Battery: {phone.get_battery()}%")
```
